```python
import json
import glob
import os
import numpy as np
import pandas as pd
pd.set_option('display.max_rows', None)
```

```python
results_dir = "results/again3"
files = sorted(glob.glob(os.path.join(results_dir, "*.json")))

records = []
for file in files:
    with open(file, "r") as f:
        data = json.load(f)

    filename = os.path.basename(file).lower()

    # model name 결정
    if "wo_struct" in filename:
        model_name = "scl w/o struct"
    elif "wo_potential" in filename:
        model_name = "scl w/o potential"
    elif "wo_semantic" in filename:
        model_name = "scl w/o semantic"
    elif "scl_seed" in filename:
        model_name = "scl"
    else:
        model_name = data["model"].lower()

    row = {
        "file": os.path.basename(file),
        "model": model_name,
        "dataset": data.get("dataset"),
        "seed": data.get("seed"),
        "training_time": data.get("training_time"),
        "num_parameters": data.get("num_parameters"),
    }

    for k, v in data["test_result"].items():
        row[k.lower()] = v

    records.append(row)

df = pd.DataFrame(records)
df
```

```python
df.groupby("model")[["recall@10", "ndcg@10", "precision@10", "map@10", "training_time"]].agg(["mean", "std"])
```

```python
summary = df.groupby("model")[["recall@1","recall@2","recall@3","recall@4","recall@5","recall@10","recall@20","recall@30","recall@50",
                               "precision@1","precision@2","precision@3","precision@4","precision@5","precision@10","precision@20","precision@30","precision@50",
                               "ndcg@1","ndcg@2","ndcg@3","ndcg@4","ndcg@5","ndcg@10","ndcg@20","ndcg@30","ndcg@50",
                               "map@1","map@2","map@3","map@4","map@5","map@10","map@20","map@30","map@50",
                               "training_time","num_parameters"]].agg(["mean","std"])

mean_cols = summary.xs('mean', level=1, axis=1)

summary["recall_avg"] = mean_cols[[col for col in mean_cols.columns if "recall@" in col]].mean(axis=1)
summary["recall_std"] = mean_cols[[col for col in mean_cols.columns if "recall@" in col]].std(axis=1)
summary["precision_avg"] = mean_cols[[col for col in mean_cols.columns if "precision@" in col]].mean(axis=1)
summary["precision_std"] = mean_cols[[col for col in mean_cols.columns if "precision@" in col]].std(axis=1)
summary["ndcg_avg"] = mean_cols[[col for col in mean_cols.columns if "ndcg@" in col]].mean(axis=1)
summary["ndcg_std"] = mean_cols[[col for col in mean_cols.columns if "ndcg@" in col]].std(axis=1)
summary["map_avg"] = mean_cols[[col for col in mean_cols.columns if "map@" in col]].mean(axis=1)
summary["map_std"] = mean_cols[[col for col in mean_cols.columns if "map@" in col]].std(axis=1)

summary
```

```python
data = pd.DataFrame(summary).T
data
```

```python
data.to_csv('fn_results4.csv', index=True)
```
